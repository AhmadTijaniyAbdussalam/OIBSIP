# DESCRIPTIVE STATISTICS

```python
numerical = df[[ 'Sales','Quantity','Discount','Profit']]
statistics = pd.DataFrame({
    'Mean': numerical.mean(),
    'Median': numerical.median(),
    'Mode': numerical.mode().iloc[0],
    'Standard deviation': numerical.std(),
    'Variance': numerical.var(),
    'Minimum': numerical.min(),
    'Maximum': numerical.max(),

})
statistics
```

OBSERVATION :-
The descriptive statistics reveal that sales are highly skewed, with a few high-value orders significantly increasing the average sales figure. Customers typically purchase arround 3-4 items per transaction,indicating relatively stable purchasing quantities. Discounts vary considerably, with 20% being a common discount level. profitability,however, is incosistent, as many transactions generate low or zero profit while some result in significant looses. The large difference between average and median profit also suggest that a view highly profitable transaction influences overall performance.

## THE SERIES ANALYSIS

```python
df['Order Date'] = pd.to_datetime(df['Order Date'],errors='coerce')
```

```python
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

OBSERVATION :- Monthly Sales show a clear upward trend from 2014 to 2017, with increasing peaks over time. Sales are highly seasonal, with strong performance in the later year. The highest sales occured in the late 2017, exceeding $120,000,indicating significant business growth and improving performance over the four-year period.

```python
quarterly_sales = df.groupby(df['Order Date'].dt.to_period('Q'))['Sales'].sum()
quarterly_sales
```

```python
plt.figure(figsize=(12, 6))
plt.plot(quarterly_sales.index.astype(str), quarterly_sales.values, marker='o')
plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

OBSERVATION :-
Quartely sales reveal powerful seasonal pattern with Q4 consistently delivers the highest revenue every year, peaking dramatically in 2017Q4 to about $280,000. While sale dip in Q1,the  overall trends shows strong growth from 2014 to 2017, proving the business is expanding successfully with year-end period.

## PRODUCT ANALYSIS

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Superstore.csv", encoding="latin1")
```

```python
top_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)
top_products
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))
top_products.sort_values(ascending=True).plot(kind='barh', color='steelblue')
plt.title('Top 10 Products by Sales',fontsize=14)
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.tight_layout()
plt.show()
```

OBSERVATION :-  
The Canon imageCLASS 2200 Advanced Copier dominates sales, generating significantly higher revenue than any other product. Most of the top 10 products are high-value office machines Copier,binding systems, and printers. This shows that a small number of premium technology and office equipment items drive a large portion of total sales, highlighting strong demand for durable business machines.

```python
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
category_sales
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12, 6))
category_sales.plot(kind='bar', color='steelblue')
plt.title('Sales by Category',fontsize=14)
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

OBSERVATION :-  
The Technology among others category generate significantly high amount of about $836,000. This shows that Technology contribute the large portion to the Total revenue

# HEAT MAP

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
numerical = df[['Sales','Quantity','Discount','Profit']]
```

```python
correlation = numerical.corr()
correlation
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Numerical Variables')
plt.tight_layout()
plt.show()
```

OBSERVATION :-  
Sales and Profit ave moderate positive correlation(0.48), showing that higher sales generally increases profit. However, Discount has a weak negative correlation with profit(-0.22),meaning higher discount tends to reduce profitability. Quantity shows almost no meaningful relationship with sales or Profit. Overall, sales volume(revenue) drives profit more than quantity or discounting.
