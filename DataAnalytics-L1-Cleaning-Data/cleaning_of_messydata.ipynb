# DATA CLEANING

MISSING DATA TREATMENT

```python
numeric_columns = df.select_dtypes(include=np.number).columns
```

```python
df[numeric_columns].skew()
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
df['tenure_years'] = df['tenure_years'].fillna(df['tenure_years'].mean())
df['years_experience'] = df['years_experience'].fillna(df['years_experience'].mean())
df['performance_review_score'] = df['performance_review_score'].fillna(df['performance_review_score'].median())
```

```python
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
df['education_level'] = df['education_level'].fillna(df['education_level'].mode()[0])
df['remote_status'] = df['remote_status'].fillna(df['remote_status'].mode()[0])
df['performance_rating'] = df['performance_rating'].fillna(df['performance_rating'].mode()[0])
df['ethnicity'] = df['ethnicity'].fillna('unknown')
```

JUSTIFICATION:-

Different imputation strategies were applied based on the data type,distribution,and amount of missing data in each column. such that, in;


* Numerical columns such as Age, Tenure year, years of experience and Performance Review Score: Mean imputation was used because these variables were not highly skewed, making the mean the reasonable representation of the typical value with the exception of Performance review Score which median was used because it was highly skewed, and the median is less affected by extreme values and therefore provide more reliable replacement for the missing observation.

* Categories columnms such as Gender, Educational Level, Remote status, and Performance Rating: Mode imputation was used because these variables are categorical, and the most frequently occuring category provides a simple and appropriate replacement for the relatively small number of missing values

* For ethnicity ; Missing values were replaced with "Unknown" because the column contain a relatively large number of missing observations. Rather than asssigning these missing records to a particular ethnic group without evidence,"Unknown"preserves the original uncertainty and avoiding introducing bias into analysis.

STANDARDIZATION OF INCOSISTENT FORMATTING

```python
df['gender'] = df['gender'].replace({
    'Man': 'Male',
    'M': 'Male',
    'Woman': 'Female',
    'F': 'Female',
    'NB': 'Non-Binary'

})
```

```python
df['gender'] = df['gender'].str.title()
df['gender'] = df['gender'].str.strip()
```

```python
df['education_level'] = df['education_level'].replace({
    "bachelor's degree": "Bachelor's",
    "Bachelors'": "Bachelor's",
    'Bachelors': "Bachelor's",
    'BSc': "Bachelor's",
    'MSc': "Master's",
    'PhD.': 'PhD',
    "Master'S": "Master's",
    "Bachelor'S": "Bachelor's"
})
```

```python
df['education_level'] = df['education_level'].str.title()
df['education_level'] = df['education_level'].str.strip()
```

```python
df['department'] = df['department'].str.title()
df['department'] = df['department'].str.strip()
```

```python
df['remote_status'] = df['remote_status'].replace({
    'On site': 'Onsite',
    'On-site': 'Onsite',
    'Meets Expectational': 'Meets',
    'Exceeds Expectations': 'Exceeds'
})
```

```python
df['remote_status'] = df['remote_status'].str.title()
df['remote_status'] = df['remote_status'].str.strip()
```

DATA TYPE CORRECTION

```python
df['employee_id'] = df['employee_id'].astype('string')
```

```python
df['hire_date'] = df['hire_date'].astype(str)
```

```python
df['salary_raw'] = df['salary_raw'].str.replace(r'[^\d.-]', '', regex=True).astype(float)
```

```python
df['salary_raw'] = pd.to_numeric(df['salary_raw'], errors='coerce')
df['salary_raw'] = df['salary_raw'].fillna(df['salary_raw'].median())
```

OUTLIER DETECTION

```python
for col in numeric_columns:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
  print(f"{col}:{len(outliers)} outliers")
```

```python
for col in numeric_columns:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_bound = Q1 - 1.5 * IQR
  upper_bound = Q3 + 1.5 * IQR
  outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)][col]
  print(f"\n{col}")
  print(outliers.tolist())
```

The IQR method were used to identify potential outliers in the numerical variables. The detected outlier in age, tenure years, salary were retained because the value were considered to be realistic and potentially meaningful observations rather than data-entry errors. Therefore, no capping or removal was performed.
